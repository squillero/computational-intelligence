```python
from itertools import permutations
import numpy as np
```

```python
TICTACTOE_MAP = np.array([[1, 6, 5], [8, 4, 0], [3, 2, 7]])


def display(x, o):
    for r in range(3):
        for c in range(3):
            if TICTACTOE_MAP[r, c] in x:
                print("X", end=" ")
            elif TICTACTOE_MAP[r, c] in o:
                print("O", end=" ")
            else:
                print(".", end=" ")
        print()
```

```python
def won(cells):
    return any(sum(h) == 12 for h in permutations(cells, 3))


def eval_terminal(x, o):
    if won(x):
        return 1
    elif won(o):
        return -1
    else:
        return 0
```

```python
def minmax(board):
    # board: ({my cells}, {my opponent's cells})

    # Value of the position as it is, before any further move
    val = eval_terminal(*board)

    # Free cells: all cells minus those taken by either player
    possible = list(set(range(9)) - board[0] - board[1])

    # Leaf: someone has already won, or the board is full (draw)
    if val != 0 or not possible:
        return None, val

    evaluations = list()
    for ply in possible:
        # Play `ply`, then pass the turn: the opponent becomes the player to
        # move. Its best value from there is the value of `ply` for us,
        # with the sign flipped (a win for the opponent is a loss for us).
        _, val = minmax((board[1], board[0] | {ply}))
        evaluations.append((ply, -val))

    # Pick the move with the highest value; among equal values, the first
    # one found (no preference for quicker wins)
    return max(evaluations, key=lambda k: k[1])
```

```python
p = ({1, 3}, {5, 8})
display(*p)
print()
best_ply, eval = minmax(p)
display(p[0] | {best_ply}, p[1])
```
