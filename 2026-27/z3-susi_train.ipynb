```python
# Both the RiSEs' z3 binary and the z3-solver Python package are needed.
# see: https://en.wikipedia.org/wiki/Z3_Theorem_Prover
import z3
```

```python
solver = z3.Solver()

c = z3.Int("c")
m = z3.Int("m")

# Start with 6 cherries, pick up 1 every 200m, eat 2 every 150m
solver.add(c == 6 + m / 100 - 2 * m / 150)
# just ate the last two cherries
solver.add(c == 0)
solver.add(m % 150 == 0)

z3.pp(solver.assertions())
if solver.check() == z3.sat:
    print("Found a solution: ", end="")
    z3.pp(solver.model())
```
