```python
from random import random, choice
from icecream import ic
```

```python
NUM_OBJECTS = 10
MAX_WEIGHT = 100
VALUE = {o: random() * 10 for o in range(NUM_OBJECTS)}
WEIGHT = {o: random() * 10 for o in range(NUM_OBJECTS)}
```

```python
def fitness(solution: dict[int, float]) -> float:
    value = sum(n * VALUE[o] for o, n in solution.items())
    if sum(n * WEIGHT[o] for o, n in solution.items()) > MAX_WEIGHT:
        return 0.0
    else:
        return value


def tweak1(solution: dict[int, float]) -> dict[int, float]:
    k1 = choice(list(solution.keys()))
    k2 = choice(list(solution.keys()))
    new_solution = dict(solution)
    new_solution[k1] = solution[k2]
    new_solution[k2] = solution[k1]
    return new_solution


def tweak(solution: dict[int, float]) -> dict[int, float]:
    k = choice(list(solution.keys()))
    new_solution = dict(solution)
    if random() < 0.5:
        new_solution[k] += 1
    elif new_solution[k] > 0:
        new_solution[k] -= 1
    return new_solution
```

```python
solution = {o: 0.0 for o in range(NUM_OBJECTS)}
steps = 0
ic(steps, fitness(solution))
while True:
    steps += 1
    new_solution = tweak(solution)
    if fitness(new_solution) > fitness(solution):
        solution = dict(new_solution)
        ic(steps, fitness(solution))
ic(steps, fitness(solution))
None
```
