```python
from random import choice, randint
from collections.abc import Sequence
from icecream import ic
```

```python
DIMENSION = 100
```

```python
def fitness(solution: list[bool]) -> int:
    return sum(solution)

def tweak(solution: list[bool]) -> list[bool]:
    i = randint(0, DIMENSION-1)
    return solution[:i] + [not solution[i]] + solution[i+1:]
```

```python
solution = [choice([True, False]) for _ in range(DIMENSION)]
```

```python
solution = [choice([True, False]) for _ in range(DIMENSION)]
steps = 0
ic(steps, fitness(solution))
while fitness(solution) < DIMENSION:
    steps += 1
    new_solution = tweak(solution)
    if fitness(new_solution) > fitness(solution):
        ic(steps, fitness(new_solution))
        solution = new_solution[:]
ic(steps, fitness(solution))
None
```
